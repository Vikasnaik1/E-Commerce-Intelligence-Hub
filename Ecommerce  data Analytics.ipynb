```python
 
# create analytical dataset

df = Orders.merge(
    customers,
    on="customer_Id",
    how="left"
)
```

```python
# 2. Add OrderDetails
df = df.merge(
    Orderdetails,
    on="Order_ID",
    how="inner"
)
```

```python
df = df.merge(
    products,
    left_on="product_ID",
    right_on="Product_Id",
    how="left"
)
```

```python
df. shape
```

```python
df.duplicated().sum()
```

```python
df.drop(columns=["Product_Id"], inplace=True)
```

```python
df. shape
```

```python
df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month
df["MonthName"] = df["OrderDate"].dt.month_name()
df["Quarter"] = df["OrderDate"].dt.quarter
df["Day"] = df["OrderDate"].dt.day
df["OrderDayName"] = df["OrderDate"].dt.day_name()
df["YearMonth"] = df["OrderDate"].dt.to_period("M").astype(str)
```

```python
# profit margin
df["Profit_Margin"] = (df["Profit"] / df["SalesAmount"]) * 100
```
